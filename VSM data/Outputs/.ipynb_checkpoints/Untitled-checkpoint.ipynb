# Main Research Questions
## 1. Can rivers form a meandering planform in the absence of plants?
a. Answer is yes, and our field example is a rare example. We will investigate how river hydraulics and sediment transport produce a stable single-threaded channel that actively meanders. This makes the Quinn River a rare analog to rivers on the early Earth or Mars.

## 2. What sets river bank erosion rates in the absence of plants?
a. Classic ideas suggest that banks need to be strong to meander. What is causing this
strength in the case of the Quinn River? Is it salt crusts? Or mud? How do migration
rates compare to erosion theory?

## 3. What are the curvilinear textures on the floodplain and how do they record river migration?
<B>a.</b> Most floodplains on Earth are covered in vegetation, making it difficult to decipher floodplain architecture without trenching or geophysical surveys. The Quinn river, however, shows beautiful curvilinear textures. These features appear similar to deposit textures visible in orbital imagery on Mars. So this field site may offer a rare Earth analog to understand the origin of these textures.
<B>b.</b> Aretheyabandonedlevees(scrollbars)whicharecommoninvegetatedmeanders? Or are they bedding planes from later accretion of point bars? If the later, how did they come to be exposed/truncated at the surface? Can they be used as traces of former positions of the river? How do their curvatures relate to the active channel bend curvature?

## 4. What the heck is a fluvial ridge and how did it form?
<B>a.</b> Another rare landform apparent at the Quinn River is a fluvial ridge. A fluvial ridge
is a topographic ridge that appears similar to a river channel in planform, but, unlike a channel, it has positive relief. These landforms are observed across Mars and their origin is uncertain. Often the ridge is thought to be a channel cast. For example, if the channel was filled with a resistant material, like lava, then differential erosion results in the channel fill material standing up as a ridge.
<B>b.</b> What material composes the ridge and is it stronger than the neighboring material? How did it come to form? Do its dimensions (width, relief, top slope) reflect active channel dimensions?

## 5. What is the timescale required for the building up of an unvegetated channel belt?
<B>a.</b> Both early Mars and Earth exhibit evidence of channel belts, with examples of Mars being associated with fluvial ridges. The Quinn River stands out due to its exceptional preservation, particularly the remarkable transitions we observe across diverse features like multiple generations of meander belts, fluvial ridges, and active channels. We will collect samples for optically stimulated luminescence and radiocarbon dating from these features could provide insights into the formation
and evolution of the entire channel belt over time.

```python
import matplotlib.pyplot as plt
import numpy as np

# Define lithologies, facies, grain sizes, paleocurrent directions, and bioturbation levels
lithologies = ['Sandstone', 'Siltstone', 'Shale']
facies = ['Channel', 'Floodplain', 'Point Bar']
grain_sizes = ['Coarse', 'Medium', 'Fine']
paleocurrents = ['E', 'SE', 'S']
bioturbation = ['Low', 'Moderate', 'High']

# Define depths (in meters)
depths = np.linspace(0, 100, 11)

# Create a figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Plot lithologies
for lithology, depth in zip(lithologies, depths):
    ax.plot([0, 1], [depth, depth], color='k', lw=2)
    ax.text(-0.1, depth, lithology, ha='right', va='center')

# Plot facies, grain sizes, paleocurrents, and bioturbation
for i, depth in enumerate(depths):
    facies_index = i % len(facies)
    grain_size_index = i % len(grain_sizes)
    paleocurrent_index = i % len(paleocurrents)
    bioturbation_index = i % len(bioturbation)
    
    ax.text(1.1, depth, facies[facies_index], ha='left', va='center')
    ax.text(1.3, depth, grain_sizes[grain_size_index], ha='left', va='center')
    ax.text(1.5, depth, paleocurrents[paleocurrent_index], ha='left', va='center')
    ax.text(1.7, depth, bioturbation[bioturbation_index], ha='left', va='center')

# Add scale
ax.plot([0.5, 0.5], [0, 100], color='k', lw=1)
ax.text(0.5, -5, 'Scale', ha='center', va='top')
ax.text(0.5, -10, '(m)', ha='center', va='top')

# Add notes
ax.text(-0.1, 105, 'Notes:', ha='right', va='center')
ax.text(0.5, 105, 'This is a stratigraphic section', ha='center', va='center')

# Set axis limits and labels
ax.set_xlim(-0.2, 2)
ax.set_ylim(-15, 115)
ax.set_xlabel('Lithology / Facies / Grain Size / Paleocurrent / Bioturbation')

# Remove y-axis ticks and spines
ax.yaxis.set_ticks([])
ax.spines['top'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['right'].set_visible(False)

# Show plot
plt.title('Stratigraphic Section')
plt.grid(axis='x')
plt.show()
```
